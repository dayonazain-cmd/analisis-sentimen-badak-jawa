```python
import os
import pandas as pd
from googleapiclient.discovery import build
from googleapiclient.errors import HttpError
from dotenv import load_dotenv
from IPython.display import display

# Memuat API Key 
load_dotenv()

API_KEY = os.getenv('YOUTUBE_API_KEY')
# Inisialisasi 
youtube = build('youtube', 'v3', developerKey=API_KEY)
```

```python
# Video: "5 risiko yang bisa terjadi kalau badak Jawa dipindahkan"
VIDEO_ID = "f9k1WOIvahE"
MAKSIMAL_KOMENTAR = 100

print(f"Target Video ID: {VIDEO_ID}")
print(f"Batas maksimal komentar yang akan diambil: {MAKSIMAL_KOMENTAR}")
```

```python
data_komentar = []

try:
    # Mengambil data komentar dari YouTube API
    response = youtube.commentThreads().list(
        part='snippet',
        videoId=VIDEO_ID,
        maxResults=MAKSIMAL_KOMENTAR,
        order='time'
    ).execute()

    # Mengekstrak informasi penting dari respons JSON
    for item in response['items']:
        komentar = item['snippet']['topLevelComment']['snippet']
        data_komentar.append({
            'video_id': VIDEO_ID,
            'komentar': komentar['textDisplay'],
            'tanggal': komentar['publishedAt']
        })
        
    print(f"Proses pengambilan selesai. Total komentar didapat: {len(data_komentar)}")

except HttpError as e:
    print(f"Terjadi kesalahan {e}")
```

```python
# Mengubah list data menjadi DataFrame pandas
df_mentah = pd.DataFrame(data_komentar)

# Menampilkan 5 baris pertama untuk memastikan data sudah benar
display(df_mentah.head(10))

# Menyimpan data ke dalam file CSV
nama_file = 'dataset_badak_jawa.csv'
df_mentah.to_csv(nama_file, index=False)

print(f"Data berhasil disimpan ke dalam file: {nama_file}")
```
