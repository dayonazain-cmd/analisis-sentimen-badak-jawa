# **3. Data Preprocessing**

```python
import pandas as pd
import re
import html
from IPython.display import display

# Memuat dataset (menggunakan hasil dari scraping)
df_prep = pd.read_csv('dataset_badak_jawa.csv')

# Menghapus nilai kosong (jika ada) sebelum diproses
df_prep = df_prep.dropna(subset=['komentar'])

# Fungsi untuk membersihkan teks
def bersihkan_teks(teks):
    # Mengubah ke huruf kecil
    teks = str(teks).lower()
    
    # Mengubah entitas HTML seperti &quot; menjadi karakter asli
    teks = html.unescape(teks)
    
    # Menghapus URL
    teks = re.sub(r'http\S+|www\S+|https\S+', '', teks, flags=re.MULTILINE)
    
    # Menghapus tag HTML seperti <br>
    teks = re.sub(r'<.*?>', '', teks)
    
    # Menghapus karakter selain huruf (a-z) dan spasi
    teks = re.sub(r'[^a-z\s]', '', teks)
    
    # Menghapus spasi berlebih
    teks = re.sub(r'\s+', ' ', teks).strip()
    
    return teks
```

```python
# Menerapkan fungsi ke dataset
df_prep['komentar_bersih'] = df_prep['komentar'].apply(bersihkan_teks)

# Menghapus baris yang menjadi kosong setelah dibersihkan
df_prep = df_prep[df_prep['komentar_bersih'] != '']

print("Perbandingan Teks Sebelum dan Sesudah Preprocessing:")
display(df_prep[['komentar', 'komentar_bersih']].head(10))

# menyimpan dataset final
nama_file_bersih = 'dataset_badak_jawa_clean.csv'
df_prep.to_csv(nama_file_bersih, index=False)

print(f"\nData bersih berhasil disimpan ke dalam file: {nama_file_bersih}")
```
