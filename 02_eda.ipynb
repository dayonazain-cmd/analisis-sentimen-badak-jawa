# **2. Exploratory Data Analysis (EDA)**

```python
import pandas as pd
from IPython.display import display

# Memuat dataset hasil scraping
df_eda = pd.read_csv('dataset_badak_jawa.csv')

print(f"Jumlah data awal: {len(df_eda)} baris\n")

# Cek data kosong
print("Jumlah data kosong per kolom:")
print(df_eda.isnull().sum())
print("\n")

# Cek dan hapus duplikat
duplikat = df_eda.duplicated(subset=['komentar']).sum()
print(f"Jumlah komentar duplikat: {duplikat} baris")

if duplikat > 0:
    df_eda = df_eda.drop_duplicates(subset=['komentar'])
    print(f"Data duplikat dihapus. Sisa data: {len(df_eda)} baris.")
```

```python
# Membuat kolom baru berisi jumlah kata per komentar
df_eda['jumlah_kata'] = df_eda['komentar'].apply(lambda x: len(str(x).split()))

print("Statistik Panjang Komentar (dalam jumlah kata):")
display(df_eda['jumlah_kata'].describe())
```
